```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/sample_data/emma.jpg", cv2.IMREAD_GRAYSCALE)

def intensity_transform(pixel):
    if pixel < 50:
        return int((50/50) * pixel)
    elif pixel < 100:
        return int(((100-50)/(100-50)) * (pixel-50) + 50)
    elif pixel < 150:
        return int(((255-100)/(150-100)) * (pixel-100) + 100)
    else:
        return int(((255-150)/(255-150)) * (pixel-150) + 150)


vectorized_transform = np.vectorize(intensity_transform)
transformed_img = vectorized_transform(img)

plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Transformed Image")
plt.imshow(transformed_img, cmap='gray')
plt.axis('off')

plt.show()
```
